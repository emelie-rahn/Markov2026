```python
import numpy as np
```

```python
for m in range(3, 13):
    P = np.zeros((m, m)) # transition matrix

    for i in range(m - 1):
        P[i, i + 1] = 1

    P[m - 1, 0] = 0.5
    P[m - 1, 2] = 0.5

    A = (P > 0).astype(int)
    B = np.eye(m, dtype = int)

    bound = (m - 1) ** 2 + 1 # wielandt's bound
    M = None

    for n in range(1, bound + 1):
        B = ((B @ A) > 0).astype(int)

        if B.all():
            M = n
            break

    eigenvalues = np.linalg.eigvals(P)
    moduli = np.sort(np.abs(eigenvalues))[::-1] # sorted from largest to smallest
    lambda2 = moduli[1]

    print(
        f"m = {m}, "
        f"M = {M}, "
        f"Wielandt bound = {bound}, "
        f"|lambda_2| = {lambda2:.4f}"
    )
```
