```python
import numpy as np

# num of simulations
N_values = [100, 500, 1000, 5000, 10000, 50_000, 100_000, 500_000, 1_000_000, 5_000_000]
estimates = []

for N in N_values: 
    X = np.random.uniform(0, 1, N)
    Y = np.random.uniform(0, 1, N)
    Z = np.random.uniform(0, 1, N)

    condition1 = X ** 2 + Y ** 2 < Z
    condition2 = Z ** 2 > X * Y

    success = condition1 & condition2
    estimate = np.mean(success)
    estimates.append(estimate)
    
print(estimates)
```

```python
import matplotlib.pyplot as plt

plt.plot(N_values, estimates, marker='o', color='blue', label='Monte Carlo Estimate')

plt.axhline(y = 0.3763, color = 'green', linestyle='--', label='Theoretical Baseline (0.3763)')

plt.xscale('log')

plt.xlabel('Sample size N')
plt.ylabel('Estimated probability')
plt.title('Monte Carlo Estimate')
plt.legend()

plt.show()
```
