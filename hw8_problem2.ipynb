```python
# parameters and libraries
import numpy as np
from scipy.stats import poisson

rng = np.random.default_rng(42)

n = 200_000
window = 5
q = 0.4
```

```python
# first way
star_all = rng.poisson(3 * window, n)
sky_all = rng.poisson(2 * window, n)

star_thin = rng.binomial(star_all, q)
sky_thin = rng.binomial(sky_all, q)
total_thin = star_thin + sky_thin
```

```python
# second way
star_direct = rng.poisson(1.2 * window, n)
sky_direct = rng.poisson(0.8 * window, n)
total_direct = star_direct + sky_direct

def summarize(name, star, sky, total):
    print(f"\n{name}")
    print("Star mean / variance:",
          star.mean(), star.var(ddof=1))
    print("Sky mean / variance:",
          sky.mean(), sky.var(ddof=1))
    print("Total mean / variance:",
          total.mean(), total.var(ddof=1))
    print("Star-sky correlation:",
          np.corrcoef(star, sky)[0, 1])

summarize("Thin after generating", star_thin,
          sky_thin, total_thin)
summarize("Generate detected directly", star_direct,
          sky_direct, total_direct)
```

```python
# compare total counts with Poisson(10)
print("\nTotal count distribution")
for k in range(16):
    empirical = np.mean(total_direct == k)
    theoretical = poisson.pmf(k, 10)
    print(k, empirical, theoretical)

mask = total_direct == 10
p_exactly_6 = np.mean(star_direct[mask] == 6)
print("\nWindows with 10 detections:", mask.sum())
print("P(exactly 6 star | total=10):", p_exactly_6)
```

```python
# first detection time and type
ts = rng.exponential(scale=1/1.2, size=n)
tb = rng.exponential(scale=1/0.8, size=n)

star_first = ts < tb
M = np.minimum(ts, tb)

print("\nFirst detection")
print("P(star first):", star_first.mean())
print("P(star first and within 1 sec):",
      np.mean(star_first & (M <= 1)))
print("Mean first detection time:", M.mean())
print("Correlation of time and type:",
      np.corrcoef(M, star_first.astype(int))[0, 1])
```
