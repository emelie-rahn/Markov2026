Problem 5, part c

```python
mu = 1.0
N_max = 10000
R = 1000
C = np.ones(R, dtype=float)

for n in range(3, N_max):
    p = mu * C / n
    u = np.random.uniform(0, 1, R)
    C += (u < p)

z_empirical = C / N_max

empirical_mean = np.mean(z_empirical)
empirical_std = np.std(z_empirical)
empirical_cv = empirical_std / empirical_mean

min_core = int(np.min(C))
max_core = int(np.max(C))

plt.figure(figsize = (9, 6))

plt.hist(z_empirical, bins = 40, density = True, alpha = 0.6, label = 'Vectorized Simulation')

z_vals = np.linspace(0, 1, 500)
h_z = 2 * (1.0 - z_vals)
plt.plot(z_vals, h_z, label='Theoretical h(z) = 2(1-z)')

plt.xlabel('Core Density')
plt.ylabel('Probability Density')
plt.title('Limiting Density h(z) vs Network Growth (mu = 1)')
plt.xlim(0, 1)
plt.legend()
plt.show()
```
